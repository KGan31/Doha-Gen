```python
import re

def clean_poem(text):
    
    # Remove extra slashes and author repeats
    text = re.sub(r"/.*", "", text)
    
    # Fix missing spaces after punctuation
    text = re.sub(r"([।,!?])([^\s])", r"\1 \2", text)
    
    # Add line breaks at sentence boundaries
    text = text.replace("।", "।\n")
    
    # Remove multiple spaces
    text = re.sub(r"\s+", " ", text)
    
    # Remove leading/trailing spaces
    text = text.strip()
    
    return text
```

```python
from datasets import load_dataset

dataset = load_dataset(
    "csv",
    data_files="/kaggle/input/datasets/anonymous111111111/hindi-poem-dataset/kavitas_cleaned.csv"
)

def process(example):
    text = example["kavita_text"]
    clean = clean_poem(text)
    return {"text": clean}

dataset = dataset.map(process)
```

```python
dataset = dataset.remove_columns(
    ["poet_name", "kavita_name", "kavita_text", "kavita_url"]
)
```

```python
def filter_fn(example):
    return len(example["text"]) > 50

dataset = dataset.filter(filter_fn)
```
